# Part 1.

```python
import pandas as pd
import numpy as np
from io import StringIO
```

## 0. LOADING DATA (read_csv)

```python
# In the real test you'll write: df = pd.read_csv('data.csv')
# Here we read from an in-memory string so the script is self-contained.
# Note the intentional messiness: blanks (-> NaN), 'unknown', mixed case,
# a price stored with a stray space, a duplicate row.
 
raw = """id,name,dept,salary,hire_date,active
1,Alice,Sales,50000,2020-01-15,yes
2,Bob,sales,,2019-06-01,yes
3,Charlie,Engineering,80000,2021-03-10,no
4,Dana,ENGINEERING,82000,2020-11-20,yes
5,Eve,Sales,55000,unknown,no
6,Frank,Marketing, 60000 ,2022-05-05,yes
7,Grace,Engineering,90000,2018-09-30,yes
7,Grace,Engineering,90000,2018-09-30,yes
8,Heidi,Marketing,,2021-07-14,no
"""
 
# key read_csv args you should know: na_values converts given tokens to NaN
df = pd.read_csv(StringIO(raw), na_values=["unknown", ""])
print("Loaded DataFrame:\n", df)
```

## 1. INSPECTING

```python
# .head / .tail — first / last N rows (default 5)
print(".head(3):\n", df.head(3))
 
# .shape — (rows, columns)
print("\n.shape ->", df.shape)
 
# .dtypes — data type of each column. Note salary is float (NaNs force float),
# and hire_date/active are 'object' (strings) — we'll fix these.
print("\n.dtypes:\n", df.dtypes)
 
# .info() — dtypes + non-null counts in one view (prints, returns None)
print("\n.info():")
df.info()
 
# .describe() — summary stats for NUMERIC columns
print("\n.describe():\n", df.describe())
 
# .columns / .index — the labels
print("\n.columns ->", list(df.columns))
print(".index   ->", list(df.index))
 
# .nunique — count of distinct values per column
print("\n.nunique():\n", df.nunique())
 
# .unique — the distinct values in ONE column
print("\ndf['dept'].unique() ->", df["dept"].unique())
 
# .value_counts — frequency of each value (normalize=True gives proportions)
print("\ndf['dept'].value_counts():\n", df["dept"].value_counts())
print("\nvalue_counts(normalize=True):\n", df["dept"].value_counts(normalize=True))
```

## 2. SELECTING COLUMNS & ROWS

```python
# Single column -> a Series
print("df['name'] (a Series):\n", df["name"])
 
# Multiple columns -> a DataFrame (note the double brackets)
print("\ndf[['name','salary']]:\n", df[["name", "salary"]])
 
# .loc[rows, cols] — LABEL based (here labels happen to be 0..8)
print("\n.loc[0:2, ['name','dept']]:\n", df.loc[0:2, ["name", "dept"]])
 
# .iloc[rows, cols] — POSITION based (integer positions, end exclusive)
print("\n.iloc[0:3, 0:2]:\n", df.iloc[0:3, 0:2])
```

## 3. FILTERING ROWS (boolean masking)

```python
# Single condition
print("salary > 60000:\n", df[df["salary"] > 60000])
 
# Multiple conditions — & (and) / | (or), each condition in ()
print("\nEngineering AND salary>80000:\n",
      df[(df["dept"] == "Engineering") & (df["salary"] > 80000)])
 
# .isin — membership test against a list
print("\ndept in [Sales, Marketing]:\n",
      df[df["dept"].isin(["Sales", "Marketing"])][["name", "dept"]])
 
# .between — inclusive range test
print("\nsalary between 55k and 82k:\n",
      df[df["salary"].between(55000, 82000)][["name", "salary"]])
 
# .query — filter with a readable string expression
print("\n.query('salary > 80000'):\n",
      df.query("salary > 80000")[["name", "salary"]])
```

## 4. MISSING VALUES

```python
# .isna() / .isnull() are identical — True where value is NaN
print("df.isna():\n", df.isna())
 
# Count NaNs per column: chain .isna().sum()
print("\nNaNs per column:\n", df.isna().sum())
 
# % missing per column
print("\n% missing:\n", (df.isna().mean() * 100).round(1))
 
# .dropna — remove rows with any NaN (subset limits which cols to check)
print("\ndropna(subset=['salary']) removes salary-NaN rows:\n",
      df.dropna(subset=["salary"])[["name", "salary"]])
 
# .fillna — replace NaN. Here: fill salary with its median.
median_sal = df["salary"].median()
filled = df.copy()
filled["salary"] = filled["salary"].fillna(median_sal)
print(f"\nfillna salary with median ({median_sal}):\n",
      filled[["name", "salary"]])
```

## 5. FIXING DTYPES & CLEANING TEXT

```python
clean = df.copy()
 
# The 'salary' for Frank had a stray space (" 60000 ").
# read_csv already parsed salary as float, but if it were text you'd do:
#   pd.to_numeric(col, errors='coerce') turns bad values into NaN instead of crashing
print("pd.to_numeric on a messy column:")
messy_numbers = pd.Series(["10", " 20 ", "oops", "40"])
print("  input   ->", list(messy_numbers))
print("  coerced ->", list(pd.to_numeric(messy_numbers, errors="coerce")))
 
# pd.to_datetime — parse strings into real dates (errors='coerce' -> NaT on bad)
clean["hire_date"] = pd.to_datetime(clean["hire_date"],errors="coerce", dayfirst = True, )
print("\nhire_date after to_datetime:\n", clean[["name", "hire_date"]])
print("hire_date dtype now ->", clean["hire_date"].dtype)
 
# .str methods — clean inconsistent text. dept has 'sales','Sales','ENGINEERING'
clean["dept"] = clean["dept"].str.strip().str.title()
print("\ndept after .str.strip().str.title():\n", clean["dept"].unique())
 
# .map — replace values via a dictionary (e.g. yes/no -> 1/0)
clean["active"] = clean["active"].map({"yes": 1, "no": 0})
print("\nactive after .map(yes->1,no->0):\n", clean[["name", "active"]].head())
 
# .replace — swap specific values (works on whole df or a column)
print("\n.replace example (Marketing->Mktg):\n",
      clean["dept"].replace({"Marketing": "Mktg"}).unique())
 
# .astype — force a dtype (only safe once NaNs are handled)
clean["active"] = clean["active"].astype("int64")
print("\nactive dtype after astype(int64) ->", clean["active"].dtype)
 
```

## 6. DUPLICATES

```python
# .duplicated — True for rows that repeat an earlier row (Grace appears twice)
print("duplicated():\n", df.duplicated())
print("\nnumber of duplicate rows ->", df.duplicated().sum())
 
# .drop_duplicates — remove them (keep='first' by default)
deduped = df.drop_duplicates()
print("shape before ->", df.shape, " after drop_duplicates ->", deduped.shape)
```

## 7. RENAMING, DROPPING, CLIPPING, BINNING

```python
# .rename — change column names via a dict
renamed = df.rename(columns={"salary": "annual_salary"})
print("columns after rename ->", list(renamed.columns))
 
# .drop — remove columns (axis=1) or rows
dropped = df.drop(columns=["active"])
print("columns after drop 'active' ->", list(dropped.columns))
 
# .clip — cap values into a [lower, upper] range (outlier handling)
print("\nsalary clipped to [55000, 85000]:\n",
      df["salary"].clip(lower=55000, upper=85000).tolist())
 
# pd.cut — bin a continuous column into labelled ranges
sal_no_na = filled["salary"]
bands = pd.cut(sal_no_na, bins=[0, 60000, 80000, 100000],
               labels=["low", "mid", "high"])
print("\npd.cut salary into bands:\n",
      pd.DataFrame({"salary": sal_no_na, "band": bands}))
 
# pd.qcut — bin into equal-sized QUANTILE groups (here 3 -> terciles)
print("\npd.qcut into 3 quantile groups:\n",
      pd.qcut(sal_no_na, q=3).value_counts())
```

# Part 2.

```python
sales = pd.DataFrame({
    "region":  ["North", "North", "South", "South", "North", "South", "North", "South"],
    "rep":     ["Alice", "Bob", "Cara", "Dan", "Alice", "Cara", "Bob", "Dan"],
    "product": ["A", "B", "A", "B", "B", "A", "A", "B"],
    "units":   [10, 5, 8, 12, 7, 15, 9, 4],
    "revenue": [1000, 400, 800, 1200, 560, 1500, 900, 400],
    "date":    pd.to_datetime([
        "2023-01-05", "2023-01-12", "2023-01-15", "2023-02-01",
        "2023-02-10", "2023-02-20", "2023-03-03", "2023-03-15"]),
})

print("Sales table:\n", sales)
```

## 1. GROUPBY — split / apply / combine

```python
# Simplest: one group key, one aggregate
print("revenue by region (mean):\n",
      sales.groupby("region")["revenue"].mean())
 
# .size() counts rows per group; .sum()/.count()/.mean() aggregate
print("\nrows per region (.size()):\n", sales.groupby("region").size())
 
# Multiple aggregates on multiple columns via .agg with a dict
print("\n.agg per region:\n",
      sales.groupby("region").agg(
          total_rev=("revenue", "sum"),
          avg_units=("units", "mean"),
          n_orders=("revenue", "count"),
      ))
 
# Group by TWO keys
print("\nrevenue by region & product:\n",
      sales.groupby(["region", "product"])["revenue"].sum())
```

## 2. transform vs agg  (KEY DISTINCTION)

```python
# .agg collapses each group to ONE row.
# .transform returns a result the SAME LENGTH as the input — perfect for
# adding a group statistic back onto every original row.
 
sales2 = sales.copy()
sales2["region_avg_rev"] = sales2.groupby("region")["revenue"].transform("mean")
sales2["pct_of_region"] = (sales2["revenue"] / sales2["region_avg_rev"] * 100).round(1)
print("transform broadcasts group mean back to each row:\n",
      sales2[["region", "rep", "revenue", "region_avg_rev", "pct_of_region"]])
```

## 3. apply — custom per-group / per-row logic

```python
# apply a custom function to each group (here: revenue spread within region)
spread = sales.groupby("region")["revenue"].apply(lambda s: s.max() - s.min())
print("revenue spread (max-min) per region:\n", spread)
 
# apply row-wise (axis=1) to build a new column
sales_apply = sales.copy()
sales_apply["price_per_unit"] = sales_apply.apply(
    lambda row: round(row["revenue"] / row["units"], 1), axis=1)
print("\nrow-wise apply -> price per unit:\n",
      sales_apply[["rep", "revenue", "units", "price_per_unit"]])
 
```

## 4. SORTING & TOP-N

```python
# .sort_values — order by a column
print("sorted by revenue desc:\n",
      sales.sort_values("revenue", ascending=False)[["rep", "revenue"]])
 
# .nlargest / .nsmallest — fast top/bottom N without full sort
print("\ntop 3 orders by revenue:\n",
      sales.nlargest(3, "revenue")[["rep", "revenue"]])
 
# .rank — assign ranks (method controls tie handling)
ranked = sales.copy()
ranked["rev_rank"] = ranked["revenue"].rank(ascending=False, method="dense").astype(int)
print("\nrevenue rank:\n", ranked[["rep", "revenue", "rev_rank"]].sort_values("rev_rank"))
```

## 5. PIVOT TABLE & CROSSTAB

```python
# .pivot_table — spreadsheet-style: rows x cols -> aggregated values
pt = sales.pivot_table(index="region", columns="product",
                       values="revenue", aggfunc="sum", fill_value=0)
print("pivot_table (region x product -> sum revenue):\n", pt)
 
# margins=True adds row/column totals
pt2 = sales.pivot_table(index="region", columns="product",
                        values="revenue", aggfunc="sum",
                        fill_value=0, margins=True, margins_name="Total")
print("\nwith margins (totals):\n", pt2)
 
# pd.crosstab — frequency table of two categoricals
print("\ncrosstab region x product (counts):\n",
      pd.crosstab(sales["region"], sales["product"]))
 
```

## 6. RESHAPING — melt / pivot / stack / unstack

```python
# Start from a wide summary
wide = pt.reset_index()
print("wide format:\n", wide)
 
# .melt — wide -> long (unpivot). Turns columns into rows.
long = wide.melt(id_vars="region", var_name="product", value_name="revenue")
print("\nmelt -> long format:\n", long)
 
# .pivot — long -> wide (the inverse of melt, no aggregation)
back = long.pivot(index="region", columns="product", values="revenue")
print("\npivot -> back to wide:\n", back)
 
# .stack / .unstack — move a column level to/from the index
grouped = sales.groupby(["region", "product"])["revenue"].sum()
print("\ngrouped (MultiIndex Series):\n", grouped)
print("\n.unstack() (product -> columns):\n", grouped.unstack())
```

## 7. MERGING / JOINING TABLES

```python
#  A second table to join on 'rep' — the classic SQL JOIN in pandas.
 
targets = pd.DataFrame({
    "rep":    ["Alice", "Bob", "Cara", "Dan", "Erin"],  # Erin has no sales
    "target": [1500, 1300, 2000, 1600, 1000],
})
print("targets table:\n", targets)
 
# inner join — only reps present in BOTH tables (Erin dropped)
rep_rev = sales.groupby("rep")["revenue"].sum().reset_index()
print("\nrevenue per rep:\n", rep_rev)
 
inner = rep_rev.merge(targets, on="rep", how="inner")
print("\ninner merge (matches only):\n", inner)
 
# left join — keep ALL reps from left, NaN where right has no match
left = targets.merge(rep_rev, on="rep", how="left")
print("\nleft merge (all targets, Erin has NaN revenue):\n", left)
 
# outer join — keep everything from both
outer = rep_rev.merge(targets, on="rep", how="outer")
print("\nouter merge (union):\n", outer)
 
# pd.concat — stack DataFrames (axis=0 rows, axis=1 columns)
print("\npd.concat axis=0 (stack rows):\n",
      pd.concat([sales.head(2), sales.tail(2)])) 
```

## 8. DATES & TIME SERIES

```python
ts = sales.set_index("date").sort_index()
 
# .dt accessor — pull parts out of a datetime column
parts = sales.copy()
parts["year"] = parts["date"].dt.year
parts["month"] = parts["date"].dt.month
parts["weekday"] = parts["date"].dt.day_name()
print(".dt accessor (year/month/weekday):\n",
      parts[["date", "year", "month", "weekday"]])
 
# .resample — group by time frequency (needs a datetime index)
# 'ME' = month end. Sum revenue per month.
print("\nresample monthly revenue:\n",
      ts["revenue"].resample("ME").sum())
 
# .rolling — moving window (e.g. 2-order moving average of revenue)
print("\n2-order rolling mean of revenue:\n",
      ts["revenue"].rolling(window=2).mean())
 
# .shift — move values down (for lag features / period-over-period)
# .diff — difference vs previous row; .pct_change — % change
monthly = ts["revenue"].resample("ME").sum()
print("\nmonthly with shift/diff/pct_change:\n",
      pd.DataFrame({
          "revenue": monthly,
          "prev": monthly.shift(1),
          "diff": monthly.diff(),
          "pct_change": (monthly.pct_change() * 100).round(1),
      }))
 
```

## 9. RESET / SET INDEX (housekeeping you do constantly)

```python
g = sales.groupby("region")["revenue"].sum()
print("groupby result (region is the index):\n", g)
 
# .reset_index — turn the index back into a normal column (very common after groupby)
print("\n.reset_index() -> region becomes a column:\n", g.reset_index())
 
# .set_index — make a column the index
print("\n.set_index('rep') on first 3 rows:\n",
      sales.head(3).set_index("rep")[["region", "revenue"]])
```
